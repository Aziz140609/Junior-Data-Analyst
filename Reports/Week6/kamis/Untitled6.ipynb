```python
import pandas as pd
```

```python
df = pd.read_csv('/content/Salary Data.csv')
display(df.head())
df.info()
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix: Age, Years of Experience, Salary')
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

df_encoded = df.copy()
le = LabelEncoder()
df_encoded['Gender_Encoded'] = le.fit_transform(df['Gender'])
df_encoded['Education_Encoded'] = le.fit_transform(df['Education Level'])

corr_matrix_full = df_encoded.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix_full, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix (Termasuk Kategorikal yang Di-encode)')
plt.show()
```
